# 3.1 Problem - 1A - Single Year HDI Exploration (Latest Year: 2022)

```python
# Import libraries for data processing and visualization

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

**1. Extract Latest Year:**

```python
# Read the HDI dataset from a CSV file

df = pd.read_csv("Human_Development_Index_Dataset.csv", encoding = "ISO-8859-1")
```

```python
# Display all unique years present in the dataset

df["year"].unique()
```

```python
# Create a separate DataFrame containing HDI data for the year 2022

hdi_2022_df = df[df["year"]==2022].copy()
```

**2. Data Exploration:**

```python
# Show the first 10 records for the year 2022

print("Initial 10 Data for year 2022")
hdi_2022_df.head(10)
```

```python
# View structural information of the 2022 HDI dataset

hdi_2022_df.info()
```

**3. Missing Values & Data Cleaning:**

```python
# Display missing values per column and the total missing value count

display(hdi_2022_df.isnull().sum())
print ("Total Count : ")
print(hdi_2022_df.isnull().sum().sum())
```

```python
# Identify categorical and numerical columns in the dataset

categorical_columns = hdi_2022_df.select_dtypes(include=['object']).columns
numerical_columns = hdi_2022_df.select_dtypes(exclude=['object']).columns
```

Replacing the special characters with Nan for consistent calcuation.

```python
# Standardize missing values by replacing symbols with NaN

hdi_2022_df.replace(["–", "-", "NA", "n/a"], np.nan, inplace=True)
```

Appropriate respective conversion to numeric type and replacement with Nan if missing or malformed data.

```python
# Ensure numerical columns contain valid numeric values

for col in numerical_columns:
    hdi_2022_df[col] = pd.to_numeric(hdi_2022_df[col], errors='coerce')
```

```python
# Strip extra spaces from the 'country' column

hdi_2022_df['country'] = hdi_2022_df['country'].str.strip()
```

```python
# Drop duplicate records

hdi_2022_df.drop_duplicates(inplace=True)
```

```python
# Drop rows with missing HDI values

hdi_2022_df = hdi_2022_df.dropna(subset=['hdi'])
```

Imputing with mean if missing values exceed 5% in overall else dropping the rows.

```python
# Handle missing values in numerical columns (except 'hdi'):
# - Calculate the percentage of missing values for each column
# - If missing values > 5%, fill them with the column mean
# - If missing values ≤ 5%, drop the rows containing missing values
# This ensures data completeness while minimizing data loss

for col in numerical_columns:
    if col == 'hdi':
        continue
    missing_pct = hdi_2022_df[col].isnull().sum() / len(hdi_2022_df) * 100
    if missing_pct > 5:
        hdi_2022_df[col] = hdi_2022_df[col].fillna(hdi_2022_df[col].mean())
    else:
        hdi_2022_df = hdi_2022_df.dropna(subset=[col])
```

```python
# Show the top rows of the cleaned dataset

print("Cleaned Data:\n")
hdi_2022_df.head()
```

**4. Basic Statistics:**

```python
# Calculate basic statistical measures for the 'hdi' column:
# - Mean: average HDI value
# - Median: middle value of HDI
# - Standard deviation: measure of HDI variability

hdi_mean = hdi_2022_df['hdi'].mean()
hdi_median = hdi_2022_df['hdi'].median()
hdi_std = hdi_2022_df['hdi'].std()
```

```python
# Display mean, median, and standard deviation of HDI

print(f"Mean: {hdi_mean:.2f}")
print(f"Median: {hdi_median:.2f}")
print(f"Standard Deviation: {hdi_std:.2f}")
```

```python
# Display countries with the highest HDI in 2022

max_hdi = hdi_2022_df['hdi'].max()
highest_hdi_countries = hdi_2022_df[hdi_2022_df['hdi'] == max_hdi]['country'].values
print(f"Countries with highest HDI ({max_hdi}): {highest_hdi_countries}")
```

```python
# Find the minimum HDI value and identify the country or countries that have it
# Prints the country names along with their HDI score

min_hdi = hdi_2022_df['hdi'].min()
lowest_hdi_countries = hdi_2022_df[hdi_2022_df['hdi'] == min_hdi]['country'].values
print(f"Countries with lowest HDI ({min_hdi}): {lowest_hdi_countries}")
```

**5. Filtering and Sorting:**

```python
# Filter the dataset to include countries with HDI greater than 0.8
# Sort the filtered countries by Gross National Income (GNI) per capita in descending order
# Select the top 10 countries based on GNI per capita
# Display relevant columns: country, year, HDI, GNI per capita, and life expectancy


filtered_df = hdi_2022_df[hdi_2022_df['hdi'] > 0.800]
filtered_sorted = filtered_df.sort_values(by='gross_inc_percap', ascending=False)
top10_filtered = filtered_sorted.head(10)
print("Top 10 countries with (HDI > 0.8) sorted by GNI per capita\n")
top10_filtered[['country', 'year', 'hdi', 'gross_inc_percap', 'life_expectancy']]
```

6. Adding HDI Category Column:

```python

# Function to categorize HDI values into descriptive categories
# Parameters:
#   hdi : numeric value of Human Development Index
# Returns:
#   String representing HDI category: 'Low', 'Medium', 'High', or 'Very High'
# Thresholds based on standard UNDP HDI classification:
#   - Low:        < 0.550
#   - Medium:   0.550 – 0.699
#   - High:       0.700 – 0.799
#   - Very High: ≥ 0.800


def hdi_categories(hdi):
    if hdi < 0.550:
        return 'Low'
    elif 0.550 <= hdi <= 0.699:
        return 'Medium'
    elif 0.700 <= hdi <= 0.799:
        return 'High'
    else:
        return 'Very High'
```

```python
# Create a new column 'HDI_Category' by applying the hdi_categories function
# This categorizes countries based on their HDI values (e.g., Low, Medium, High, Very High)
# Display the count of countries in each HDI category
# Show a comparison of HDI and its category for the first 10 countries


hdi_2022_df['HDI_Category'] = hdi_2022_df['hdi'].apply(hdi_categories)
print(hdi_2022_df['HDI_Category'].value_counts())

print()
print("HDI with its HDI Category Comparision\n")
hdi_2022_df[['country', 'hdi', 'HDI_Category']].head(10)
```

```python
# Export the cleaned and categorized dataset to a CSV file

hdi_2022_df.to_csv("HDI_category_added.csv", index=False)
```

**2. Composite Development Score:**

```python
# Read the South Asia HDI CSV file into a DataFrame

hdi_south_asia_df = pd.read_csv("HDI_SouthAsia.csv").copy()
```

```python
# Select HDI data for South Asian countries for the year 2022

hdi_south_asia_2022 = hdi_south_asia_df[hdi_south_asia_df['year'] == 2022].copy()
```

```python
# Clean and preprocess the South Asia HDI data for 2022:
# 1. Replace common missing value indicators ("–", "-", "NA", "n/a") with NaN
# 2. Convert key numerical columns ('hdi', 'life_expectancy', 'gross_inc_percap') to numeric, coercing invalid entries to NaN
# 3. Remove duplicate rows to ensure unique records
# 4. Handle missing values:
#    - If missing percentage > 5%, fill with column mean
#    - If missing percentage ≤ 5%, drop the rows containing missing values


hdi_south_asia_2022.replace(["–", "-", "NA", "n/a"], np.nan, inplace=True)
numeric_cols = ['hdi', 'life_expectancy', 'gross_inc_percap']

for col in numeric_cols:
    hdi_south_asia_2022[col] = pd.to_numeric(hdi_south_asia_2022[col], errors='coerce')

hdi_south_asia_2022.drop_duplicates(inplace=True)

for col in numeric_cols:
    missing_pct = hdi_south_asia_2022[col].isnull().sum() / len(hdi_south_asia_2022) * 100

    if missing_pct > 5:
        hdi_south_asia_2022[col] = hdi_south_asia_2022[col].fillna(hdi_south_asia_2022[col].mean())
    else:
        hdi_south_asia_2022 = hdi_south_asia_2022.dropna(subset=[col])
```

```python
# Compute a composite score using 30% life expectancy and 30% GNI per capita

hdi_south_asia_2022['Composite_Score'] = (0.30 * hdi_south_asia_2022['life_expectancy'] + 0.30 * hdi_south_asia_2022['gross_inc_percap'])
```

```python
# Assign ranks based on composite score (higher score = better rank)

hdi_south_asia_2022['Composite_Rank'] = hdi_south_asia_2022['Composite_Score'].rank(ascending=False)
```

```python
# Get top 5 countries by composite score

top5_composite = hdi_south_asia_2022.sort_values('Composite_Score', ascending=False).head(5)
```

**3. Outlier Detection:**

```python
# Calculate the Interquartile Range (IQR) and identify potential outlier thresholds
# for 'hdi' and 'gross_inc_percap' in South Asian countries for 2022
# - Q1: 25th percentile, Q3: 75th percentile
# - IQR = Q3 - Q1
# - Lower and upper bounds for detecting outliers = Q1 - 1.5*IQR, Q3 + 1.5*IQR

Q1_hdi = hdi_south_asia_2022['hdi'].quantile(0.25)
Q3_hdi = hdi_south_asia_2022['hdi'].quantile(0.75)
IQR_hdi = Q3_hdi - Q1_hdi

lower_hdi = Q1_hdi - 1.5 * IQR_hdi
upper_hdi = Q3_hdi + 1.5 * IQR_hdi

Q1_gni = hdi_south_asia_2022['gross_inc_percap'].quantile(0.25)
Q3_gni = hdi_south_asia_2022['gross_inc_percap'].quantile(0.75)
IQR_gni = Q3_gni - Q1_gni

lower_gni = Q1_gni - 1.5 * IQR_gni
upper_gni = Q3_gni + 1.5 * IQR_gni
```

```python
# Detect outliers in the 2022 South Asia HDI dataset
# - A row is marked as an outlier if:
#    * 'hdi' is below lower_hdi or above upper_hdi
#    * OR 'gross_inc_percap' is below lower_gni or above upper_gni
# - Add a new column 'Outlier' (True if outlier, False otherwise)
# - Display the country, HDI, GNI per capita, and outlier status

hdi_south_asia_2022['Outlier'] = (
    (hdi_south_asia_2022['hdi'] < lower_hdi) | (hdi_south_asia_2022['hdi'] > upper_hdi) |
    (hdi_south_asia_2022['gross_inc_percap'] < lower_gni) | (hdi_south_asia_2022['gross_inc_percap'] > upper_gni)
)

print("Outlier Detection (HDI & GNI per Capita) for South Asia, 2022:\n")
print(hdi_south_asia_2022[['country','hdi','gross_inc_percap','Outlier']])
```

```python
# Scatter plot to visualize HDI vs GNI per capita for South Asian countries in 2022
# - Points are colored based on outlier status (red = outlier, blue = normal)
# - Helps identify which countries deviate significantly in HDI or GNI per capita

plt.figure(figsize=(8,6))

sns.scatterplot(data=hdi_south_asia_2022, x='gross_inc_percap', y='hdi', hue='Outlier', palette={False: 'blue', True: 'red'},)
plt.title("HDI vs GNI per Capita (South Asia, 2022)")
plt.xlabel("GNI per Capita")
plt.ylabel("Human Development Index (HDI)")
plt.legend(title="Outlier")
plt.show()
```

Using the 1.5*IQR rule, no countries in 2022 are classified as outliers in either HDI or GNI per capita because it lies between the lower and upper bound for both.

**7. Outlier Detection:**

```python
# Function to detect outliers in a numeric column using the IQR method
# Parameters:
#   df  : pandas DataFrame
#   col : column name (string) in which to detect outliers
# Returns:
#   DataFrame containing only the rows that are outliers in the specified column
# Method:
#   - Calculate Q1 (25th percentile) and Q3 (75th percentile)
#   - Compute IQR = Q3 - Q1
#   - Define outlier thresholds: lower = Q1 - 1.5*IQR, upper = Q3 + 1.5*IQR
#   - Return rows where column values are below lower or above upper threshold

def detect_outliers(df, col):
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR
    outliers = df[(df[col] < lower) | (df[col] > upper)]
    return outliers
```

```python
# Detect and display outliers in South Asia for HDI and GNI per Capita (2020–2022)
# - Uses the previously defined detect_outliers() function
# - Prints the countries and values that are considered outliers
# - Helps identify unusually low or high HDI and GNI per Capita values

# South Asia
outlier_hdi_south = detect_outliers(hdi_south_asia_2020_2022, 'hdi')
outlier_gni_south = detect_outliers(hdi_south_asia_2020_2022, 'gross_inc_percap')

print("South Asia HDI Outliers:")
print(outlier_hdi_south[['country','hdi']])

print("\nSouth Asia GNI per Capita Outliers:")
print(outlier_gni_south[['country','gross_inc_percap']])
```

```python
# Scatter plot showing HDI vs GNI per Capita for South Asia (2020–2022)
# - Blue points: Normal countries (non-outliers)
# - Red points: HDI outliers detected using IQR method
# - Orange points: GNI per Capita outliers detected using IQR method
# - Highlights unusual countries in terms of HDI or income
# - Figure saved as 'hdi_SA_outliers.png' at 300 DPI

plt.figure(figsize=(8,6))

plt.scatter(hdi_south_asia_2020_2022['gross_inc_percap'], hdi_south_asia_2020_2022['hdi'], color='blue', label='Normal Countries')

plt.scatter(outlier_hdi_south['gross_inc_percap'], outlier_hdi_south['hdi'], color='red', label='HDI Outliers')

plt.scatter(outlier_gni_south['gross_inc_percap'], outlier_gni_south['hdi'], color='orange', label='GNI Outliers')

plt.xlabel("GNI per Capita")
plt.ylabel("HDI")
plt.title("South Asia: HDI vs GNI per Capita")
plt.legend()
plt.savefig('hdi_SA_outliers', dpi=300)
plt.show()
```

No HDI and GNI outlier exists in the HDI vs GNI per capita of South Asia (2020 -2022)
